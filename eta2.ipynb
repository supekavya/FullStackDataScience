```python
import pandas as pd
df=pd.read_csv('loan_approved.csv')


mean_val = df['ApplicantIncome'].mean()
std_val = df['ApplicantIncome'].std()

print("Mean:", mean_val)
print("Standard Deviation:", std_val)
```

```python
df1=pd.get_dummies(df['Property_Area'],prefix='Property_Area',drop_first=True)
df=pd.concat([df,df1],axis=1).drop(['Property_Area'],axis=1)
```

```python
df2=pd.get_dummies(df['Gender'],prefix='Gender',drop_first=True)
df=pd.concat([df,df2],axis=1).drop('Gender',axis=1)
```

```python
df3=pd.get_dummies(df['Education'],prefix='Education',drop_first=True)
df=pd.concat([df,df3],axis=1).drop('Education',axis=1)
```

```python
df4=pd.get_dummies(df,columns=['Married'],prefix='Married',drop_first=True)
df=pd.concat([df,df4],axis=1).drop('Married',axis=1)    
```

```python
df5=pd.get_dummies(df['Self_Employed'],prefix='Self_Employed',drop_first=True)
df=pd.concat([df,df5],axis=1).drop('Self_Employed',axis=1)
```

```python
df = df.loc[:, ~df.columns.duplicated()]
print((df.columns))
```
