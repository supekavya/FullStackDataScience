Naive Bayes Machine Learning Algorithm implementation

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
```

```python
# import iris dataset
from sklearn.datasets import load_iris
x, y = load_iris(return_X_y=True)
```

```python
x
```

```python
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
from sklearn.naive_bayes import GaussianNB
model=GaussianNB()
model.fit(x_train,y_train)
y_pred = model.predict(x_test)


```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
acc_score = accuracy_score(y_test,y_pred)
print(acc_score)
```

```python
confusion_matrix = confusion_matrix(y_test,y_pred)
confusion_matrix
```

```python
classication_report = classification_report(y_test,y_pred)
print(classication_report)
```

```python
from sklearn.naive_bayes import BernoulliNB
bern = BernoulliNB()
bern.fit(x_train,y_train)
y_pred_bernoulli = bern.predict(x_test)
acc_scoree = accuracy_score(y_test,y_pred_bernoulli)
print(acc_scoree)


```

```python
classication_report_bernoulli = classification_report(y_test,y_pred_bernoulli)
print(classication_report_bernoulli)
```

```python
confusion_matrix_bernoulli = confusion_matrix(y_test,y_pred_bernoulli)
print(confusion_matrix_bernoulli)
```
