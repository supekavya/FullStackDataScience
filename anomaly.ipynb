```python
import pandas as pd
import numpy as np
from sklearn.ensemble import IsolationForest
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# read the dataset
data = pd.read_csv('embedded_system_network_security_dataset.csv')
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
df.info()
```

```python
# check missing values
df.isnull().sum()
```

## Step 2: Data Preprocessing

```python
# if columns is bool type then convert in to int tyle
for col in df.columns:
    if df[col].dtype == 'bool':
        df[col] = df[col].astype(int)
```

## Feature Engineering

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_features = scaler.fit_transform(df)
```

```python
scaled_features_df = pd.DataFrame(scaled_features, columns=df.columns)
scaled_features_df.head()
```

```python
# instantiate the isolation forest model
n_estimators = 100
contamination = 0.04
sample_size = 256
```

```python
#train the isolation forest model
model = IsolationForest(n_estimators=n_estimators, contamination=contamination, max_samples=sample_size, random_state=42)
model.fit(scaled_features_df)

anomaly_labels = model.predict(scaled_features_df)

binary_labels = np.where(anomaly_labels == -1, 0, 1)
scaled_features_df['anomaly'] = binary_labels
```

```python
scaled_features_df['anomaly'].value_counts()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(x=scaled_features_df['packet_size'], y=scaled_features_df['packet_count_5s'], hue=scaled_features_df['anomaly'])
plt.title('Anomaly Detection using Isolation Forest')
plt.xlabel('Packet Size')
plt.ylabel('Packet Count in 5s')
plt.legend(title='Anomaly', loc='upper right')
plt.show()


    
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(x=scaled_features_df['inter_arrival_time'], y=scaled_features_df['src_port'], hue=scaled_features_df['anomaly'])
plt.title('Anomaly Detection using Isolation Forest')
plt.xlabel('Inter Arrival Time')
plt.ylabel('Source Port')
plt.legend(title='Anomaly', loc='upper right')
plt.show()
```
