# Matplotlib - Python Vizualisation Library

## What is matplotlib? 

Matplotlib is python library for creating static, animated and interactive visualizations

#### Why we need it? 

It helps to understand trends, patterns and outliers

## Where we use it?

Data Analysis

Machine Learning

Business purposes like to analyse market trends, finance and scientific research





## Importing Matplotlib

import matplotlib.pyplot as plt

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [10,20,25,30,40]
 
plt.plot(x, y)
plt.title("Line Plot Example")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
categories = ['A','B','C','D']
values = [10,15,7,12]
plt.plot(x,y)
plt.xlabel("categories")
plt.ylabel("values")
plt.bar(categories, values, color='green')
plt.title("Bar Plot Example")
plt.show()
```
